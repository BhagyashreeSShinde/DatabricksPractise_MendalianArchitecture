```python
#Dropdown widget for environment selection
dbutils.widgets.dropdown(
    name="environment",
    defaultValue="dev",
    choices=["dev", "prd", "qa"],
    label="select Environment"
)

#Get selected environment
env = dbutils.widgets.get("environment")
print(env)

#Define dynamic table name and source file location
brznTablName = f"saleslake_{env}.bronze_{env}.rawCustomer"
srcFileLoc   = f"/Volumes/saleslake_{env}/bronze_{env}/vol_saleslake_src_files_{env}/dailyCustomer/"

```

```python
 #COPY INTO Bronze table
spark.sql(f"""
COPY INTO  {brznTablName} 
FROM(
   SELECT customer_id,customer_name,email,phone,address,city,state,country,zip_code,segment,current_timestamp() AS ingest_ts
   FROM '{srcFileLoc}'
)
FILEFORMAT=CSV
FORMAT_OPTIONS('header'='true')
COPY_OPTIONS('mergeSchema'='true')
""")
```
